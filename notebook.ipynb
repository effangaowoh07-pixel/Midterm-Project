```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = 'https://raw.githubusercontent.com/effangaowoh07-pixel/Midterm-Project/refs/heads/main/Data/Train%20(1).csv'
```

```python
df = pd.read_csv(data)
df.head().T
```

```python
df.columns = df.columns.str.lower()

categorical_columns = list(df.dtypes[df.dtypes == 'object'].index)

for c in categorical_columns:
    df[c] = df[c].str.lower()
```

```python
# 1. Select the text columns
text_cols = df.select_dtypes(include=['string']).columns

# 2. Convert them to object
df[text_cols] = df[text_cols].astype(object)
```
